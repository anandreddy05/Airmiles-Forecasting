### Import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.graphics.tsaplots import month_plot,quarter_plot,plot_acf,plot_pacf
from statsmodels.tsa.seasonal import seasonal_decompose
```

Load The dataset

```python
df = pd.read_csv("../data/preprocessedData.csv",index_col="Date",parse_dates=True)
df.head()
```

### Plot Y

```python
df['y'].plot()
plt.savefig("../visualisations/df_y.png")
```

## Seasonality

### Monthly Plot

```python
month_plot(df["y"].asfreq("MS"))
plt.savefig("../visualisations/month_plot.png")
plt.show()
```

### Quater plot

```python
quarter_plot(df["y"].asfreq("QS"))
plt.savefig("../visualisations/quater_plot.png")
plt.show()
```

## Seasonal Decomposition

```python
decomposition = seasonal_decompose(df['y'],model='additive',period=12)
fig = decomposition.plot()
fig.set_size_inches(12, 6)
plt.savefig('../visualisations/decomposition.png')
plt.show()
```

## AutoCorrelation

```python
figs,ax = plt.subplots(figsize=(12,6))
plot_acf(df['y'],lags=100,ax=ax)
plt.savefig("../visualisations/autoCorr.png")
plt.show()
```

## Partial AutoCorrelation 

```python
figs,ax = plt.subplots(figsize=(12,6))
plot_pacf(df['y'],lags=50,ax=ax)
plt.savefig("../visualisations/PartialAutoCorr.png")
plt.show()
```

## Import librares

```python
import pandas as pd
```

### Import Data from data folder
- Set Date as the Index and parse_dates=True to convert to datetime
- Our dataset contains dates in the "DD/MM/YYYY" format instead of the default "MM/DD/YYYY" used in some systems

```python
df = pd.read_csv("../data/airmiles.csv",index_col="Date",parse_dates=True,dayfirst=True)
df.head()
```

```python
df.info()
```

### There is no frequency at df.index so we used asfreq('MS') as we are having monthly data

```python
df = df.asfreq('MS')
df.index
```

### Check for nan values

```python
df.isna().sum().sum()
```

```python
df.shape
```

### Renaming Dependent Variable(airmiles) to y

```python
df.rename(columns={"airmiles":"y"},inplace=True)
```

```python
df.head()
```

```python
import os
df.to_csv(os.path.join("..","data","preprocessedData.csv"),index=True)
```
